# Data Cleaning

```python
df.isnull().sum()
```

```python
df[df.isnull().any(axis=1).sort_index(ascending=True)]
```

```python
df.iloc[426] = df.iloc[426].fillna('Ayrton Senna')
```

```python
df.iloc[426]
```

```python
df.iloc[446] = df.iloc[446].fillna(df.iloc[446]['Director'])
df.iloc[518] = df.iloc[518].fillna(df.iloc[518]['Director'])
df.iloc[534] = df.iloc[534].fillna('Uma Thurman')

# df.loc[534, 'Main Actor(s)'] = 'Uma Thurman'
# More practical

df.iloc[556] = df.iloc[556].fillna('Matt Damon')
df.iloc[592] = df.iloc[592].fillna('Alex Honnold')
df.iloc[593] = df.iloc[593].fillna('Sixto Rodriguez')
df.iloc[611] = df.iloc[611].fillna('Cat') # a Cat
df.iloc[671] = df.iloc[671].fillna('Tilikum') #an Orca
df.iloc[696] = df.iloc[696].fillna("Ric O'Barry")

#Google all of this, i know its not practical but its fun and doable since its small missing value and not in a hurry situation
```
